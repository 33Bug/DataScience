```python
import pandas

data = pandas.read_csv("datasets/supermarket_data.csv")
data
```

```python
# data summary
data.describe()
#quantitative e.g. 7 oranges, 5(numbers)
#qualitative e.g. male,female,yellow,green(quality/description)
```

```python
# fill empty item price
data["itemprice"] = data["itemprice"].fillna(55)

data.describe()
```

```python
# generate visualizations based on the data

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.countplot(x = "gender" , data = data, palette="dark")
```

```python
sns.countplot(x = "product", data = data, palette="dark")
plt.xticks(rotation = 45)
```

```python
sns.countplot(x = "product" , hue="gender", palette="Reds", data = data)
plt.xticks(rotation = 45)
```

```python
data["payment"].value_counts().plot.pie(autopct='%1.1f%%')
```

```python
sns.countplot(x="customertype", data = data, palette="dark")
```

```python
# calculate average rating by city and customer type(grouped by chart)
plt.figure(figsize=(8,5))
sns.barplot(x="city", y="rating", hue="customertype", data =data, palette="Set2")
plt.title("Average rating by City and Customer Type")
plt.ylabel("Average rating")
plt.ylim(0, 10)
```

```python
# heat map of payment method by City
sns.heatmap(pandas.crosstab(data["city"], data["payment"]), annot=True, fmt="d" )
plt.title("Payment Method by City")
```
